Find Duplicate Values

```python
def find_duplicates(num):
    seen = {}
    set_duplicates = set()
    for i in num:
        if i not in seen:
            seen[i] = True
        else:
            set_duplicates.add(i)
    return set_duplicates

#Time Complexity = O(N)
#Space Complexity = 0(N)  - It can be 2N (set and dictionary) but constants are dropped
```

Find the second largest number in the list

```python
def second_largest_num(nums):
    large = float('-inf')
    second_large = float('-inf')
    for num in nums:
        if num > large:
            second_large = large
            large = num
        elif num > second_large and num != large:
            second_large = num
    return second_large

#Time Complexity = O(N) - looks at each number exactly once - 1 for loop
#Space Complexity = O(1)- only store two numbers in memory
```

Rotating the List

```python
def rotate_list(lst, k):
    last = []
    front = []
    p = len(lst) - k
    for el in range(p,(len(lst))):
        last.append(lst[el])
    for el in range(0, p):
        front.append(lst[el])
    rotated_list = last + front
    return rotated_list
    
```

Two Sum

```python
def two_sum(nums, target):
    left = 0
    right =  len(nums) - 1
    while left < right:
        sum = nums[left] + nums[right]
        if sum == target:
            return left, right
        elif sum > target:
            right-= 1

        elif sum < target:
            left += 1

    else:
        print("No match found")
```

reverse List (Two pointers)

```python
def reverse_list(lst):
    left = 0
    right = len(lst) -1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]
        left += 1
        right -= 1
    return lst
```

```python
def is_palindrome(text):
    left = 0
    right = len(text) - 1
    while left < right:
        if text[left] != text[right]:
            return False
        else:
            right -= 1
            left += 1
    return True
```

Browsing History - Undo Redo Logic - Stack

```python
class BrowseHistory:
    def __init__(self):
        self.current_page = "google.com"
        self.history = []
        self.forward =[]
    def visit(self, url):
        self.history.append(self.current_page)
        self.current_page = url
        self.forward = []
    def back(self):
        self.forward.append(self.current_page)
        self.current = self.history.pop()
        return self.current

    def forward(self):
        self.history.append(self.current_page)
        self.current = self.forward.pop()
        return self.current

```

```python
def backspace_string(text):
    stack = []
    for char in text:
        if char == "#":
            if stack:
                stack.pop()
        else:
            stack.append(char) 
    return "".join(stack)

text = "abc#de#f"
backspace_string(text)
```

Balanced Bracket Validator - Stack

```python
def is_balanced(code):
    open_brackets = ['(', '[', '{']
    closing_brackets =[')', ']', '}']
    brackets =[]
    for char in code:
        if char in open_brackets:
            brackets.append(char)
        if char in closing_brackets:
            if len(brackets) == 0:
                return False
            top_bracket = brackets.pop()
            if char == ')' and top_bracket != '(':
                return False
            elif char == ']' and top_bracket != '[':
                return False
            elif char == '}' and top_bracket != '{':
                return False
    return len(brackets) == 0
    
```

Queue Implementations

```python
from collections import deque
class Queue:
    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        self._items.append(item)
        print(f"Enqueued: {item}")

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Cannot dequeue from an empty queue")
        item = self._items.popleft()
        print(f"Dequeued: {item}")
        return item

    def front(self):
        if self.is_empty():
            print("List is empty")
        else:
            front_el = self._items[0]
            print(f"Front: {front_el}")
            return front_el

    def is_empty(self):
        return True if len(self._items) == 0 else False
    
```

Anagram

```python
def is_anagram(w1, w2):
    w1_list = []
    w2_list = []
    for char in w1:
        w1_list.append(char)
    for char in w2:
        w2_list.append(char)

    if sorted(w1_list) == sorted(w2_list):
        return True
    else:
        return False

```
