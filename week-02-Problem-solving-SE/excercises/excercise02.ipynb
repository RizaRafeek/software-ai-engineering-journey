Largest number in the list

```python
def largest_number(num):
    large = num[0]
    for i in num:
        if i > large :
            large = i

    return large

#Space Complexity = O(N)
#Time Complexity = O(1)
```

Whether Number exist or not

```python
def target_present(num, target):
    for i in num:
        if i == target:
            print("Element present")
            return 
    else:
        print("Element not found")

#Time Complexity = O(N)
#Space Complexity = O(1)
```

Keeping the Count of numbers in a list

```python
def number_count(num):
    dict = {}
    for i in num:
        if i not in dict:
            dict[i] = 1
        else:
            dict[i] += 1


#Time Complexity = O(N)
#Space Complexity = O(N)
```

Find Duplicate Values

```python
def find_duplicates(num):
    seen = {}
    set_duplicates = set()
    for i in num:
        if i not in seen:
            seen[i] = True
        else:
            set_duplicates.add(i)
    return set_duplicates

#Time Complexity = O(N)
#Space Complexity = 0(N)  - It can be 2N (set and dictionary) but constants are dropped
```

Calculate Sum of a list

```python
def list_sum(num):
    sum = 0
    for i in num:
        sum += i
    return sum

#time complexity = O(N)
#space Complexity = O(1)
```
