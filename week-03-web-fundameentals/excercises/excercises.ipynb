event check-in api

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

guests = [
    {"guest_id": 1, "name": "Rahul", "vip": True, "checked_in": False},
    {"guest_id": 2, "name": "Sneha", "vip": False, "checked_in": True}
]


class guestCreate(BaseModel):
    name : str
    vip : bool
    checked_in : bool

class guestCreated(BaseModel):
    guest_id : int
    name : str
    vip : bool
    checked_in : bool

@app.post("/guests")
def create_guest(guest: guestCreate):
    guest_id = len(guests) + 1
    name = guest.name
    vip = guest.vip
    checked_in = guest.checked_in
    new_guest = {"guest_id" : guest_id, "name" : name, "vip" : vip, "checked_in": checked_in}
    guests.append(new_guest)
    return {"message":f"guest {name} with id {guest_id} added successfully"}


@app.get("/guests")
def get_all_guests():
    return guests

@app.get("/guests/{guest_id}")
def get_a_guest(guest_id : int):
    for guest in guests:
        if guest["guest_id"] == guest_id:
            return guest
    return{"error" : "Guest not found"}

@app.put("/guests/{guest_id}")
def update_guest_check_in(guest_id : int):
    for guest in guests:
        if guest["guest_id"] == guest_id:
            the_guest = guest
            the_guest["checked_in"] = True
            return the_guest
    return{"error": "Guest not found"}

@app.delete("/guests/{guest_id}")
def delete_guest(guest_id : int):
    for guest in guests:
        if guest["guest_id"] == guest_id:
            guests.remove(guest)
            return {"message":f"guest with {guest_id} removed successfully! "}
    return{"error" : "Guest not found"}        
    
```
